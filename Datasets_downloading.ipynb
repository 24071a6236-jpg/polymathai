```python
!pip install -q datasets

from datasets import load_dataset
```

```python
# Load NSL-KDD from Hugging Face

nsl_kdd = load_dataset("Mireu-Lab/NSL-KDD")

print("NSL-KDD loaded successfully!")
print(nsl_kdd)
```

```python
print("NSL-KDD train rows:", len(nsl_kdd["train"]))
print("NSL-KDD test rows :", len(nsl_kdd["test"]))

print("\nFirst training row:")
print(nsl_kdd["train"][0])
```

```python
print("NSL-KDD columns:")
print(nsl_kdd["train"].column_names)

print("\nNumber of columns:")
print(len(nsl_kdd["train"].column_names))
```

```python
# Try to load N-BaIoT from Hugging Face

try:
    nbaiot = load_dataset("codymlewis/nbaiot")

    print("N-BaIoT loaded successfully!")
    print(nbaiot)

except Exception as e:
    print("N-BaIoT loading failed.")
    print("\nError:")
    print(type(e).__name__, ":", e)
```

```python
print("N-BaIoT could not be loaded using the current Hugging Face datasets loader.")
print("Selecting ToN-IoT as the fallback dataset.")
```

```python
# Fallback dataset: ToN-IoT

toniot = load_dataset("codymlewis/TON_IoT_network")

print("ToN-IoT fallback loaded successfully!")
print(toniot)
```

```python
print("ToN-IoT train rows:", len(toniot["train"]))
print("ToN-IoT test rows :", len(toniot["test"]))

print("\nFirst training row:")
print(toniot["train"][0])
```

```python
print("ToN-IoT columns:")
print(toniot["train"].column_names)

print("\nNumber of columns:")
print(len(toniot["train"].column_names))
```

```python
print("ToN-IoT label values:")
print(set(toniot["train"]["label"]))

print("\nToN-IoT attack types:")
print(set(toniot["train"]["type"]))
```
