```python
# ============================================================
# MILESTONE 6 — DELTA MACRO-F1 + DEGRADATION SLOPE
# ============================================================

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# ------------------------------------------------------------
# 1. Get clean Macro-F1 for each Dataset + Model + Seed
# ------------------------------------------------------------

clean_df = (
    milestone5_df[milestone5_df["NoiseLevel"] == 0.0]
    [["Dataset", "Model", "Seed", "MacroF1"]]
    .rename(columns={"MacroF1": "CleanMacroF1"})
)

# ------------------------------------------------------------
# 2. Calculate Delta Macro-F1
# Delta = noisy Macro-F1 - clean Macro-F1
# ------------------------------------------------------------

milestone6_df = milestone5_df.merge(
    clean_df,
    on=["Dataset", "Model", "Seed"],
    how="left"
)

milestone6_df["DeltaMacroF1"] = (
    milestone6_df["MacroF1"]
    - milestone6_df["CleanMacroF1"]
)

# ------------------------------------------------------------
# 3. Calculate degradation slope for every
#    Dataset + Model + Seed
# ------------------------------------------------------------

slope_rows = []

for (dataset, model, seed), group in milestone6_df.groupby(
    ["Dataset", "Model", "Seed"]
):

    group = group.sort_values("NoiseLevel")

    X = group["NoiseLevel"].values.reshape(-1, 1)
    y = group["MacroF1"].values

    regression = LinearRegression()
    regression.fit(X, y)

    slope_rows.append({
        "Dataset": dataset,
        "Model": model,
        "Seed": seed,
        "DegradationSlope": regression.coef_[0]
    })

slope_df = pd.DataFrame(slope_rows)

# ------------------------------------------------------------
# 4. Average slope across the 3 seeds
# ------------------------------------------------------------

method_summary = (
    slope_df
    .groupby(["Dataset", "Model"])["DegradationSlope"]
    .agg(["mean", "std"])
    .reset_index()
    .rename(columns={
        "mean": "MeanDegradationSlope",
        "std": "StdDegradationSlope"
    })
)

# Less-negative / more-positive slope = less degradation
method_summary = method_summary.sort_values(
    ["Dataset", "MeanDegradationSlope"],
    ascending=[True, False]
)

# ------------------------------------------------------------
# 5. Display results
# ------------------------------------------------------------

print("=" * 70)
print("MILESTONE 6 — DELTA MACRO-F1 + DEGRADATION SLOPE")
print("=" * 70)

print("\nDetailed Macro-F1 + Delta Macro-F1:")
display(
    milestone6_df.sort_values(
        ["Dataset", "Model", "Seed", "NoiseLevel"]
    )
)

print("\nDegradation slope by seed:")
display(slope_df)

print("\nMethod-level degradation slope summary:")
display(method_summary)

# ------------------------------------------------------------
# 6. Save Milestone 6 results
# ------------------------------------------------------------

milestone6_df.to_csv(
    "milestone6_macro_f1_delta.csv",
    index=False
)

slope_df.to_csv(
    "milestone6_degradation_slopes_by_seed.csv",
    index=False
)

method_summary.to_csv(
    "milestone6_method_slope_summary.csv",
    index=False
)

# ------------------------------------------------------------
# 7. Verification
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MILESTONE 6 COMPLETE")
print("=" * 70)

print("Detailed rows:", len(milestone6_df))
print("Expected:", 72)

print("Slope rows:", len(slope_df))
print("Expected:", 18)

print("Method summary rows:", len(method_summary))
print("Expected:", 6)

assert len(milestone6_df) == 72
assert len(slope_df) == 18
assert len(method_summary) == 6

print("\nSaved files:")
print("✅ milestone6_macro_f1_delta.csv")
print("✅ milestone6_degradation_slopes_by_seed.csv")
print("✅ milestone6_method_slope_summary.csv")
```
